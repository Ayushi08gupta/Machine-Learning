# Sorting - 
Arrange elements in ascending order
## np.sort(a , axis = -1 , kind = None , order = None)
1. a - array 
2. axis - int or None (optional)
3. kind - quicksort , mergesort,heapsort,stable (optional)
4. order - str or list of str (optional)

```python

np.sort(a)
```

```python
b = np.random.randint(1,100,25).reshape(5,5)
b
```

```python
# In 2D array axis = 1 ----> Row wise sorting 
# axis = 0 -----> Column Wise Sorting
# By default Row Wise Sorting
print("--------------------Row Wise Sorting-------------------------------")
print(np.sort(b))
print("--------------------column Wise Sorting-------------------------------")
print(np.sort(b , axis=0))
```

```python
# Sorting in descending Order - 
np.sort(a)[::-1]
```

# np.where
returns the indices of elements in an input array where the given condition is satisfied

```python
a
```

```python
# Find all values greater than 50 
np.where(a>50)
```

```python
# Replace all values > 50 with 0
# np.where(condition , true , false)
np.where(a > 50 , 0 , a)
```

```python
# Replace all even values with 1 greater than it and all odd values by one less than it
np.where(a % 2 == 0 , a+1 , a-1)
```

# Corelation coefficient -
it shows how much the two values are corelated to each other


```python
salary = np.array([20000,30000,40000,80000,1000000])
experience = np.array([1,2,3,4,5])
```
