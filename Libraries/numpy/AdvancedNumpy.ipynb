# Numpy array vs Python lists

```python
# speed - speed of numpy is faster than list 
# list 
a = [i for i in range(10000000)]
b = [i for i in range (10000000 , 20000000)]

c = []
# Import time to check how much time is required to perform operation in a list 
import time

start = time.time()
for i in range(len(a)):
    c.append(a[i] + b[i])
print(time.time() - start)
```

```python
# numpy
import numpy as np
a = np.arange(10000000)
b = np.arange(10000000,20000000)

start = time.time()

c = a + b
print(time.time() - start)
```

```python
# memory - As such the diffrence between memory required by a list or array of int64 is not much but if we make a array of int16 or int32 or int8 it will occupy less space 

a = [i for i in range(10000000)]

import sys
sys.getsizeof(a)
```

# Advanced Indexing

```python
# Normal indexing
a = np.arange(24).reshape(6,4)
a
```

```python
a[1:,1]
```

```python
# Fancy Indexing - when finding pattern of indexing is not possible than fancy indexing is used 
print('-------------------for example we want row 1 , 3 , 4 and 6--------------------')
print(a[[0,2,3,5]])
print('-------------------for example we want column 1 , 3 , 4 ------------------------')
print(a[:,[0,2,3]])
```

```python
# Boolean Indexing - It is used when logic is required 

# create a random array of range 1 to 100 and 24 elements are selected
a = np.random.randint(1,100,24).reshape(6,4)
a
```

```python
# Find all numbers greater than 50
a[a > 50]
```

```python
# Find all numbers greater than 50 and even
a[(a > 50) & (a % 2 == 0)]
```

# BroadCasting 
The term broadcasting describes how NumPy treats arrays with different shapes during arithmetic opertaions.THe smaller array is "broadcast" across the larger array so that they have compatible shapes

```python
# Same Shape
a = np.arange(6).reshape(2,3)
b = np.arange(6,12).reshape(2,3)
print('-----------------------Array A----------------------')

print(a)
print('-----------------------Array B----------------------')
print(b)
print('-----------------------Sum----------------------')
print(a+b)
```

```python
# Different Shape
a = np.arange(8).reshape(2,4)
b = np.arange(2).reshape(3,1)
print('-----------------------Array A----------------------')

print(a)
print('-----------------------Array B----------------------')
print(b)
print('-----------------------Sum----------------------')
print(a+b)
```

## BroadCasting Rules
### 1. Make the two arrays have the same number of dimensions.
            if a numbers of dimensions of the two arrays are diffrent, add new dimensions with size 1 to the head of the array with the smaller dimension.
### 2. Make each dimension of the two arrays the same size 
        if the size of each dimension of the two arrays do not match, dimensions with size 1 are stretched to the size of the other array.
        if there is a dimension whose size is not 1 in either of the two arrays, it cannot be broadcasted, and an error is raised.
