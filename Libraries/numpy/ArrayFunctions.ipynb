```python
import numpy as np
```

# Array Functions

```python
a1 = np.random.random((3,3))
a1 = np.round(a1*100)
a1
```

```python
# max/min/sum/prod
# 0 -> col and 1 -> row
# max of all rows
print(np.max(a1 , axis = 1))
# max of all columns
print(np.max(a1 , axis = 0))
print(np.min(a1 , axis = 0))
#sum of values of a row
print(np.sum(a1 , axis = 1))
# product of all values in a array
print(np.prod(a1))
```

```python
# mean/median/std/var
np.mean(a1 , axis= 0)
```

```python
np.median(a1)
```

```python
np.std(a1)
```

```python
# Trignometric Function - sin/cos/tan/cosec/sec
np.sin(a1)
```

```python
# Dot Product - number of columns of first array should be equal to the number of rows in a second array
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(20).reshape(4,5)

np.dot(a2,a3)
```

```python
# log and exponents
print(np.log(a1))
print('_______________________________________________________________')
print(np.exp(a1))
```

```python
# round/floor/ceil
print(np.round(np.random.random((2,3))))
print('_______________________________________________________________')
print(np.floor(np.random.random((2,3))))
print('_______________________________________________________________')
print(np.ceil(np.random.random((2,3))))
```

# Indexing Slicing

```python
a1 = np.arange(10)
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)
```

```python
a1
```

```python
a1[0] 
# positive indexing is from left to right
```

```python
a1[-4]
# Negative indexing is from right to left
```

```python
a2 
```

```python
print(a2[0][3] )
# or 
print(a2[1,2])
```

```python
a3
```

```python
# to access data from 3D array check the element is in first part or second then check its row and then column
print(a3[1 , 0 , 1])
```

```python
# slicing - first value shows the index of the starting part , second value shows the ending index and third shows the space 
print(a1[2:5])
print(a1[2:5:2])
```

```python
a2
```

```python
# taking 0th row and all columns of that row
print(a2[0 ,:])
print('---------------------------------')
# taking all rows of 1 column
print(a2[: ,1] )
print('---------------------------------')
# taking element of 0th row first column
print(a2[0 ,1] )
```

```python
# taking all rows from 1 and column 1 to 2
a2[1:,1:3]
```

```python
# Taking extreme values - first colon of rows and columns show need of all rows and columns but the second colon shows the gap of 2 rows and 3 columns
a2[::2,::3]
```

```python

a2[0::1, 1::2]
```

```python
a3 = np.arange(27).reshape(3,3,3)
a3
```

```python
# it is used to access array from 3d to 2d 
a3[::2]
```

```python
# taking second row of first array
a3[0,1]
```

```python
a3[2 , 1: , 1:]
```

```python
a3[::2,0,::2]
```

# Reshaping

```python
# Transpose - Making rows as columns and columns as rows 
np.transpose(a2)
```

```python
# ravel - convert nd array into 1d array
a3.ravel()
```

# Stacking

```python
a4 = np.arange(12).reshape(3,4)
a5 = np.arange(12,24).reshape(3,4)
a5
```

```python
# Horizontal stacking - adding array after array (In Stacking all arrays must have same rows and columns)
np.hstack((a4,a5,a4))
```

```python
# vertical Stacking - adding array one below another 
np.vstack((a4,a5))
```

# Splitting

```python
# Horizontal Splitting - columns must be divisible by the number of splits you want
a4
```

```python
np.hsplit(a4,2)
```

```python
# Vertical Splitting - rows must be divisible by the number of splits you want
np.vsplit(a4,3)
```
