# What is Pandas
Pandas is a fast,powerful,flexible and easy to use open source data analysis and manipulation tool, built on top of the Python programming language

# Pandas Series
A Pandas Series is like a column in a table.It is a 1-D array holding data of any type

```python
import numpy as np
import pandas as pd
```

## Series from lists

```python
# String
country = ['India','USA','America','Pakistan','Nepal']

pd.Series(country)
```

```python
# Integers
runs = [12,34,23,12,13]

pd.Series(runs)
```

```python
# Custom Index
marks = [67,57,89,100]
subjects = ['Maths','Science','English','Hindi']

pd.Series(marks,index=subjects)
```

```python
# Setting a name
pd.Series(marks,index=subjects,name='Subject Marks')
# pd.Series(list,index{optional},name{optional})
# name -----> It is used to show the name of the series
```

## Series From dictionary

```python
marks = {
    'maths': 67,
    'english':90,
    'Science':89,
    'Hindi':100
}
marks_series = pd.Series(marks , name = 'Subjects Marks')
marks_series
```

```python
# dropna
temp = pd.Series([1,np.nan,2,3,np.nan,2,4,np.nan,4,3,np.nan,7,8,np.nan,5])
temp
```

```python
temp.dropna()
```
